```python
import csv
import re

data = open('find_the_link.csv',encoding='utf-8'))
csv_data = csv.reader(data)
data_lines = list(csv_data)
```

```python
len(data_lines)
```

```python
# find the link

found = ''
web_site = []
index = 0

for line in data_lines[:len(data_lines)]:
#     print(line)
    found = line[index]
#     print(found)
    web_site.append(found)
    index += 1
    
file_link = ''.join(web_site)
print(file_link)
```

```python
# Dowload the file
import requests

f = open('Exercise_PDF.pdf','wb') 
pdf_file = requests.get(file_link)
f.write(pdf_file.content)
f.close()
```
