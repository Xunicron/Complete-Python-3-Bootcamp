```python
# more robust version to handle Google confirmation step
# MS Copilot code 

import requests

def download_file_from_google_drive(file_id, destination):
    session = requests.Session()
    URL = "https://drive.google.com/uc?export=download"

    response = session.get(URL, params={'id': file_id}, stream=True)
    token = get_confirm_token(response)

    if token:
        params = {'id': file_id, 'confirm': token}
        response = session.get(URL, params=params, stream=True)

    save_response_content(response, destination)
    print("Download complete!")

def get_confirm_token(response):
    for key, value in response.cookies.items():
        if key.startswith('download_warning'):
            return value
    return None

def save_response_content(response, destination):
    CHUNK_SIZE = 32768
    with open(destination, "wb") as f:
        for chunk in response.iter_content(CHUNK_SIZE):
            if chunk:
                f.write(chunk)        
```

## Task Two: Download the PDF from the Google Drive link and find the phone number that is in the document. 

```python
# You should get this phone number
# 505 503 4455
```

```python
from pypdf import PdfReader, PdfWriter
```

```python
f = open('Exercise_PDF.pdf','rb')
```

```python
pdf_reader = PdfReader(f)
```

```python
pdf_reader.get_num_pages()
```

```python
# cycle through each page to find a number
import re
phone = []

for num in range(pdf_reader.get_num_pages()):
    page = pdf_reader.get_page(num)
    pdf_text = page.extract_text()
    
    found = re.search(r'(\d{3})\D(\d{3})\D(\d{4})',pdf_text)
    
    if found == None:
        print(f'Nothing found on page {num}')
    else:
        phone = found.group()
        print(f'\nPhone number {phone} found on page {num}\n')
```
