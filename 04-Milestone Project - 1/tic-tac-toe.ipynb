```python
from IPython.display import clear_output
import sys

clear_output()

def disp_board(p):

    # Setup the board based on num pad positions  
    print(f" {p[7]} | {p[8]} | {p[9]} ")
    print("-----------")
    print(f" {p[4]} | {p[5]} | {p[6]} ")
    print("-----------")
    print(f" {p[1]} | {p[2]} | {p[3]} ")
    print('\n')
    

def disp_numpad():
    
    p=['','7','8','9','4','5','6','1','2','3']
    
    print('Please use the following\nfor position inputs (numpad):\n')
    
    # Show numpad  
    disp_board(p)
    print('\n')

    
def prompt_player():
    
    player1 = ''
    
    # Prompt player 1
    while player1.upper() not in ['X','O']:
        
        player1 = input("Player 1, please pick a marker X or O: ")
       
        # Set player2
        if player1.upper() == 'X':
            player2 = 'O'
        else:
            player2 = 'X'
    
    return player1.upper(),player2.upper()



def player_input():
    
    # Ask for player input
    choice ='WRONG'
    within_range = False
    
    while choice.isdigit() == False or within_range == False:
    
        # Prompt for a numpad number
        prompt_str = "0 to quit. "+"Please enter a number (0-10): \n"
        choice = input(prompt_str)
        
        if choice.isdigit() == False:
            print("Sorry that is not a digit!")
            
        if choice.isdigit() == True:
            if int(choice) in range(0,10):
                within_range = True
            else:
                within_range = False
    
    return int(choice)


def place_token(p,pos,player):

    p[pos]=player 
    disp_board(p)
    

def check_move(board,pos,player):
    token = player
    
    # Check rows
    for row in [1,4,7]:
        if board[row]==token and board[row+1]==token and board[row+2]==token:
            print(player + ' wins!')
            sys.exit()
        
    # Check columns
    for col in [1,2,3]:
        if board[col]==token and board[col+3]==token and board[col+6]==token:
            print(player + ' wins!')
            sys.exit()

    # Check diagonals
    if board[1]==token and board[5]==token and board[9]==token:
            print(player + ' wins!')
            sys.exit()

    if board[3]==token and board[5]==token and board[7]==token:
            print(player + ' wins!')
            sys.exit()
    
    # Check no one wins
    free_flag = False
    
    for i in range(1,9):
        if board[i]==' ':
            free_flag = True
       
    if free_flag==False:
        print('Draw!')
        sys.exit()
    else:
        return
       
    

def main():
    
    # Display numpad
    disp_numpad()
    
    
    # Prompt to continue
    choice = ''
    
    while choice.upper() not in ['Y','N']:
        choice = input('Enter [y] to continue, [n] to quit: ')
    
    if choice.upper() == 'N':
        sys.exit()
    
    
    # Prompt player token selection
    [player1,player2] = prompt_player()
    
    clear_output()
    
    
    # Display player tokens
    print(f'Player 1 is {player1}')
    print(f'Player 2 is {player2}\n')

    
    # Setup blank board
    board=['',' ',' ',' ',' ',' ',' ',' ',' ',' ']
    disp_board(board)
    
    
    # Start play
    pos = 10
    player = player1
    first_player = True
 
    while pos!=0:
 
        if first_player==True:
            print(f'Player 1 place {player1} at positions [1-9]')
            player = player1 
            first_player = False
        else:
            print(f'Player 2 place {player2} at positions [1-9]')
            player = player2 
            first_player = True
 
        pos = player_input()
        
        while board[pos] in ['X','O']:
            print("Position already taken!")
            pos = player_input()
        
        clear_output()
        place_token(board,pos,player)
        check_move(board,pos,player)
        
            
    sys.exit()
```

```python
main()
```
