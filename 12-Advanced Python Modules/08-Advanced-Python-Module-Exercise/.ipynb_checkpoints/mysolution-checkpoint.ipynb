## Solution

```python
## Setup solution functions

import os
import re
import shutil
import pdb

# change dir to puzzle folder
os.chdir('/Users/kyaw_ford/Documents/Python Learning/Complete-Python-3-Bootcamp/12-Advanced Python Modules/08-Advanced-Python-Module-Exercise')

# get current directory 
cur_dir = os.getcwd()

# list directory
os.listdir()

# unzip puzzle folder
shutil.unpack_archive('unzip_me_for_instructions.zip',cur_dir,'zip')

# change dir to data folder
os.chdir('extracted_content')

# Get the current working directory and assign it to a variable
w_dir = os.getcwd()

# iterate through each directory
dir_list = os.listdir()
print('Directory contains: \n', dir_list, '\n')
```

```python
## define functions


def find_pnum(path):
    
    # open file
    #print('Working on: ' +path)
    file = open(path,'r')
    contents = file.read()
    
    # debugging
    #print(contents)
    #print('\n')

    # search for a telephone number.
    phone = re.search(r'\d{3}-\d{3}-\d{4}',contents)
    
    # close file
    file.close()
    
    return phone


def list_folders(directory):
    path_list = []
    
    for folder , sub_folders , files in os.walk(directory):
        # iterate through each subfolder
        for sub_fold in sub_folders:
            #print("\t Processing Subfolder: "+sub_fold)
            path_list.append(sub_fold)

    #pdb.set_trace()        
                
    return path_list

def list_file(directory):
    file_list = []
    
    for folder , sub_folders , files in os.walk(directory):
        # iterate through each file
        for f in files:
            #print("\t Processing File: "+f)
            file_list.append(f)

    return file_list
```

```python
## run solution

# list the folder in the extracted directory
folders = list_folders(w_dir)

for sub_folder in folders:
    f_list = list_file(sub_folder)   

    for f in f_list:
        current = sub_folder +'/' +f
        print("\t Processed: " +current)
        
        search_result = find_pnum(current)
        if search_result == None:
            pass
        else:
            p_found = search_result.group()
            found_in = 'in ' +sub_folder +'/' +f
            
print('\n')
print('found phone number ')
print(p_found)
print(found_in)
            
```
