```python
def func_one(n):
    '''
    Given a number n, returns a list of string integers
    ['0','1','2',...'n]
    '''
    return [str(num) for num in range(n)]
```

```python
def func_two(n):
    '''
    Given a number n, returns a list of string integers
    ['0','1','2',...'n]
    '''
    return list(map(str,range(n)))
```

```python
func_one(10)
```

```python
func_two(10)
```

```python
import time
```

```python
# current time before coe
start_time = time.time()
# run code
result = func_one(1000000)
# current time after code
end_time = time.time()
# elapsed time
elapsed_time = end_time - start_time
print(elapsed_time)
```

```python
# current time before coe
start_time = time.time()
# run code
result = func_two(1000000)
# current time after code
end_time = time.time()
# elapsed time
elapsed_time = end_time - start_time
print(elapsed_time)
```

```python
import timeit
```

```python
stmt = '''
func_one(100)
'''
```

```python
setup = '''
def func_one(n):
    return [str(num) for num in range(n)]
'''
```

```python
timeit.timeit(stmt,setup,number=1000000)
```

```python
stmt2 = '''
func_two(100)
'''
```

```python
setup2 = '''
def func_two(n):
    return list(map(str,range(n)))
'''
```

```python
timeit.timeit(stmt2,setup2,number=1000000)
```
