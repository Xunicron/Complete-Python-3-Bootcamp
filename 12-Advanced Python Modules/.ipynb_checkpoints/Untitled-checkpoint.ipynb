```python
from collections import Counter
```

```python
mylist = [1,1,2,2,2,2,3,3,3,4,4,4]
```

```python
Counter(mylist)
```

```python
mylist = ['a','a',10,10,10]
```

```python
Counter(mylist)
```

```python
sentence = "How many times does each word show up in this sentence with a word?"
```

```python
c = Counter(sentence.split())
```

```python
c
```

```python
c.elements()
```

```python
c.most_common()
```

```python
list(c)
```
