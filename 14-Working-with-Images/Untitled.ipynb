## Colour transparency

```python
red = Image.open('red_color.jpg')
```

```python
red.show()
```

```python
red.putalpha(0)
```

```python
red.show()
```

```python
red.putalpha(128)
```

```python
red.show()
```

```python
blue = Image.open('blue_color.jpg')
```

```python
blue.show()
```

```python
blue = blue.convert("RGB")
```

```python
blue.save("blue_color.jpg","JPEG")
```

```python
blue.putalpha(128)
```

```python
blue.show()
```

```python
red.show()
```

```python
blue.paste(im=red,box=(0,0),mask=red)
```

```python
blue.show()
```

```python
blue.save("purple_color.png")
```
