```python
class Dog():
    
    def __init__(self,name):
        self.name = name
        
    def speak(self):
        return self.name + 'says woof!'
```

```python
class Cat():
    
    def __init__(self,name):
        self.name = name
        
    def speak(self):
        return self.name + ' says mew!'
```

```python
niko = Dog('niko')
felix = Cat('felix')
```

```python
print(niko.speak())
```

```python
print(felix.speak())
```

```python
for pet in [niko,felix]:
    
    print(type(pet))
    print((pet.speak()))
```

```python
def pet_speak(pet):
    print(pet.speak())
```

```python
pet_speak(felix)
```

```python
class Animal():
    
    def __init__(self,name):
        self.name = name
        
    def speak(self):
        raise NotImplementedError('Subclass must implement this abstract method')
```

```python
class Dog(Animal):
    
    def speak(self):
        return self.name+ ' says woof!'
```

```python
fido = Dog('Fido')
```

```python
fido.speak()
```

```python
class Cat(Animal):
    
    def speak(self):
        return self.name+ ' says mew!'
```

```python
isis = Cat('Isis')
```

```python
isis.speak()
```
