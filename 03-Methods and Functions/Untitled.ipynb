```python
x = 50

def func(x):
    print(f'X is {x}')
    
    # local
    x = 200
    print(f'Locally changed {x}')
```

```python
func(x)
```

```python
x = 50

def func():
    global x
    print(f'X is {x}')
    
    # local
    x = 'New Value'
    print(f'Locally changed global x to {x}')
```

```python
func()
```
