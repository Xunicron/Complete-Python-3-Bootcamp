## Turtle Graphics - This is a common project where you create a floor of 20 x 20 squares

```python
import turtle as t
```

```python
t.home()
t.clearscreen()
```

```python
# Test moves
# Draw a box

def box(size):
    t.down()
    
    for side in range(4):
        t.forward(size)
        t.right(90)

    t.up()
    
def row_boxes(num,size):
    start_pos = t.pos()
    for n in range(num):
        box(size)
        t.up()
        t.forward(size)
        t.down()
    t.up()
    t.goto(start_pos)
        
def grid(size,rows,cols):
    start = t.pos()
    x,y = start
    for r in range(rows):
        row_boxes(cols,size)
        t.up()
        y = y-size
        t.goto(x,y)
    
    t.goto(start)

    
    
        
        
```

```python
row_boxes(10,50)
```

```python
grid(20,5,5)
```
